```python
from pyspark.sql import functions as F
from pyspark.sql.functions import col, lower, when, lit
from pyspark.sql import types as T
import calendar

# ── Helper functions ───────────────────────────────────────────────────────────
def get_dates_in_month(year, month):
    _, num_days = calendar.monthrange(year, month)
    return [int(f"{year}{month:02d}{d:02d}") for d in range(1, num_days + 1)]

def get_c360_for_month(par_month):
    snapshot_day = C360_SNAPSHOT_MAP[par_month]
    c360 = (
        spark.table("trueanalytics_data.customer360.customer360_snapshot")
        .withColumn("par_month", F.substring(col("par_day"), 1, 6).cast("int") - 1)
        .filter(
            (col("par_day") == snapshot_day) &
            (col('activated_flag') == 1)
        )
        .select(
            'msisdn',
            col('demo_gender_fog_v1_final_gender_bin').alias('gender'),
            col('demo_affluence_grid_v1_affluence_group_bin').alias('monthly_pay'),
            col('demo_age_fact_v2_final_age_num'),
            col('par_month'),
            lower(col('geog_work_location_v1_province_en_cat')).alias('hometown_province'),
            lower(col('geog_work_location_v1_district_en_cat')).alias('hometown_district'),
            lower(col('geog_work_location_v1_sub_district_en_cat')).alias('hometown_subdistrict'),
            lower(col('geog_resident_location_v1_province_en_cat')).alias('work_province'),
            lower(col('geog_resident_location_v1_district_en_cat')).alias('work_district'),
            lower(col('geog_resident_location_v1_sub_district_en_cat')).alias('work_subdistrict'),
        )
        .withColumn("demo_age_fact_v2_final_age_num", F.col("demo_age_fact_v2_final_age_num").cast(T.DoubleType()))
        .withColumn("age_range",
            F.when(F.col("demo_age_fact_v2_final_age_num") <= 10, '0-10')
            .when(F.col("demo_age_fact_v2_final_age_num").between(11, 20),  '11-20')
            .when(F.col("demo_age_fact_v2_final_age_num").between(21, 30),  '21-30')
            .when(F.col("demo_age_fact_v2_final_age_num").between(31, 40),  '31-40')
            .when(F.col("demo_age_fact_v2_final_age_num").between(41, 50),  '41-50')
            .when(F.col("demo_age_fact_v2_final_age_num").between(51, 60),  '51-60')
            .when(F.col("demo_age_fact_v2_final_age_num").between(61, 70),  '61-70')
            .when(F.col("demo_age_fact_v2_final_age_num").between(71, 80),  '71-80')
            .when(F.col("demo_age_fact_v2_final_age_num") >= 81, '>81')
            .when(F.col('demo_age_fact_v2_final_age_num').isNull(), 'U')
        )
        .withColumn('monthly_pay', F.when(F.col('monthly_pay').isNull(), F.lit('unidentified')).otherwise(F.col('monthly_pay')))
        .withColumn('gender',
            F.when(F.col('gender').isNull(), F.lit('U'))
            .when(F.col('gender') == 'female', F.lit('F'))
            .when(F.col('gender') == 'male',   F.lit('M'))
            .otherwise(F.col('gender'))
        )
        .select('msisdn', 'par_month', 'gender', 'monthly_pay', 'age_range',
                'hometown_province', 'hometown_district', 'hometown_subdistrict',
                'work_province', 'work_district', 'work_subdistrict')
    )
    return c360

# ── Config ─────────────────────────────────────────────────────────────────────
MONTH_LABELS = {
    1: 'jan', 2: 'feb', 3: 'mar', 4: 'apr',  5: 'may',  6: 'jun',
    7: 'jul', 8: 'aug', 9: 'sep', 10: 'oct', 11: 'nov', 12: 'dec',
}

# Mapping: par_month (grid) → c360 snapshot par_day
C360_SNAPSHOT_MAP = {
    202503: 20250401,
    202506: 20250701,
    202511: 20251201,
}

# ← Only line to update when adding/removing months
target_months = [(2025, 3), (2025, 6), (2025, 11)]

month_config = {
    int(f"{year}{month:02d}"): {
        'label' : f"{MONTH_LABELS[month]}_{str(year)[2:]}",
        'file'  : f"{MONTH_LABELS[month]}_report.csv",
        'dates' : get_dates_in_month(year, month),
    }
    for year, month in target_months
}

BASE_PATH = '/Volumes/int-cu/cu_project/report_1'

# ── Static data loaded once ────────────────────────────────────────────────────
df_master_file = spark.read.parquet('dbfs:/Volumes/int-cu/cu_project/master_file/h3_location_mapping_to_c360.parquet')

df_day_type = spark.read.csv("/Volumes/int-cu/cu_project/csv_file/Day_type.csv", header=True, inferSchema=True)

# ── Main loop ──────────────────────────────────────────────────────────────────
for par_month, cfg in month_config.items():

    output_path = f"{BASE_PATH}/{cfg['label']}/{cfg['file']}"
    first_date  = cfg['dates'][0]

    print(f"\n{'='*60}")
    print(f"Processing month : {par_month}")
    print(f"Output path      : {output_path}")

    # Load & cache c360 for this month only
    c360_snapshot = get_c360_for_month(par_month)

    # Load & cache full-month grid
    df_h3_month = (
        spark.table("trueanalytics_data.geo_insights.h3_geo_loc")
        .filter(
            (col('par_month') == par_month) &
            (col('province').isin('Khon Kaen', 'Roi Et'))
        )
        .select('h3_index_res9', 'province', 'district', 'sub_district',
                'par_month', 'par_day', 'par_hour', 'msisdn')
        .alias('grid')
        .join(df_master_file.alias('grid_mapping'),
              col('grid.h3_index_res9') == col('grid_mapping.h3_index_res9'),
              how='left')
        .select(
            col('par_month').alias('month'),
            col('par_day').alias('date'),
            'par_hour',
            lower(col('grid_mapping.province')).alias('province'),
            lower(col('grid_mapping.district')).alias('district'),
            lower(col('grid_mapping.subdistrict')).alias('subdistrict'),
            'msisdn'
        )
        .withColumn('time',
            F.when(F.col('par_hour') <= 5, '00:00 - 05:59')
            .when(F.col('par_hour').between(6,  11), '06:00 - 11:59')
            .when(F.col('par_hour').between(12, 17), '12:00 - 17:59')
            .when(F.col('par_hour').between(18, 23), '18:00 - 23:59')
        )
        .alias('h3')
        .join(df_day_type.alias('day_type'),
              col('h3.date') == col('day_type.date'),
              how='left')
        .select('month', 'h3.date', 'day_type', 'par_hour', 'time',
                'province', 'district', 'subdistrict', 'msisdn')
    )

    # ── Inner loop: one date at a time ─────────────────────────────────────────
    for date in cfg['dates']:
        print(f"  Processing date: {date}")

        df_h3_day = df_h3_month.filter(col('date') == date)

        # Join with c360
        df_result = (
            df_h3_day.alias('h3')
            .join(c360_snapshot.alias('c360'),
                  (col('h3.msisdn') == col('c360.msisdn')) &
                  (col('h3.month')  == col('c360.par_month')),
                  how='left')
            .select('h3.*',
                    'c360.gender', 'c360.monthly_pay', 'c360.age_range',
                    'hometown_province', 'hometown_district', 'hometown_subdistrict',
                    'work_province', 'work_district', 'work_subdistrict')
        )

        # place_type logic
        df_result = df_result.withColumn('place_type',
            when(
                (col('province') == col('hometown_province')) &
                (col('district') == col('hometown_district')) &
                (col('subdistrict') == col('hometown_subdistrict')) &
                (col('province') == col('work_province')) &
                (col('district') == col('work_district')) &
                (col('subdistrict') == col('work_subdistrict')), lit('home_work'))
            .when(
                (col('province') == col('hometown_province')) &
                (col('district') == col('hometown_district')) &
                (col('subdistrict') == col('hometown_subdistrict')), lit('home'))
            .when(
                (col('province') == col('work_province')) &
                (col('district') == col('work_district')) &
                (col('subdistrict') == col('work_subdistrict')), lit('work'))
            .otherwise(lit('thirdplace'))
        )

        # Aggregation + null fills
        df_result_group = (
            df_result
            .groupBy('month', 'date', 'day_type', 'time', 'province', 'district', 'subdistrict',
                     'gender', 'age_range', 'monthly_pay', 'place_type',
                     'hometown_province', 'hometown_district', 'hometown_subdistrict')
            .agg(F.count('msisdn').alias('count'))
            .withColumn('hometown_province',    F.when(F.col('hometown_province').isNull(),    F.lit('U')).otherwise(F.col('hometown_province')))
            .withColumn('hometown_district',    F.when(F.col('hometown_district').isNull(),    F.lit('U')).otherwise(F.col('hometown_district')))
            .withColumn('hometown_subdistrict', F.when(F.col('hometown_subdistrict').isNull(), F.lit('U')).otherwise(F.col('hometown_subdistrict')))
            .withColumn('gender',      F.when(F.col('gender').isNull(),      F.lit('U')).otherwise(F.col('gender')))
            .withColumn('age_range',      F.when(F.col('age_range').isNull(),      F.lit('U')).otherwise(F.col('age_range')))
            .withColumn('monthly_pay', F.when(F.col('monthly_pay').isNull(), F.lit('unidentified')).otherwise(F.col('monthly_pay')))
            .withColumn('count',       F.when(F.col('count') <= 25,          F.lit(25)).otherwise(F.col('count')))
            .orderBy('month', 'date', 'time', 'province', 'district', 'subdistrict',
                     'gender', 'age_range', 'monthly_pay', 'place_type',
                     'hometown_province', 'hometown_district', 'hometown_subdistrict')
        )

        # Write: overwrite on first date, union-append on subsequent dates
        if date == first_date:
            df_result_group.coalesce(1).write.csv(path=output_path, header=True, mode='overwrite')
            print(f"    Written (overwrite): {output_path}")
        else:
            df_report_before = spark.read.option("header", "true").csv(output_path)
            df_combined = df_report_before.unionAll(df_result_group)
            df_combined.coalesce(1).write.csv(path=output_path, header=True, mode='overwrite')
            print(f"    Written (append):    {output_path}")

    print(f"Month {par_month} complete.")

print("\nAll months complete.")
```
